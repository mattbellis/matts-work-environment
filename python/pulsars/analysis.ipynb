```python
import pandas as pd
import matplotlib.pylab as plt
import numpy as np

from psrqpy import QueryATNF
```

```python
# Query specific parameters (e.g., Pulsar Name, Period, and Period Derivative)
query = QueryATNF(params=['JNAME', 'RAJ', 'DecJ', 'P0', 'P1', 'F0', 'F1', 'Dist', 'XX', 'XX', 'YY', 'R_Lum', 'Age'])

# Convert the results to a pandas DataFrame for easy manipulation and plotting
df = query.table

# Display the top 5 pulsars
#print(df.head())
type(df)

df.columns
```

```python
data = df['P0']
# permutations(data, 2) creates all (a, b) pairs where a != b positionally
ratios = [a / b for a, b in itertools.permutations(data, 2) if b != 0]
```

```python
import boost_histogram as bh
```

```python
hist = bh.Histogram(bh.axis.Regular(100, 200, 210))

hist.fill(ratios)
counts = hist.view()


# 3. Visualize the histogram
fig, ax = plt.subplots(figsize=(8, 5))
ax.bar(
    hist.axes[0].centers, 
    hist.view(), 
    width=hist.axes[0].widths, 
    edgecolor='black', 
    align='center'
)

mmu = 105.658
me = 0.511

r = mmu/me

plt.axvline(r)



ax.set_title("1D Boost-Histogram with Matplotlib")
ax.set_xlabel("Value")
ax.set_ylabel("Frequency")
plt.show()
```
